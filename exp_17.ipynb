```python
    import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
import pandas as pd
data ={
    'Days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Study_Hours': [2,3,2.5,4,3.5],
    'Marks':[50,60,55,70,75],
    'Attendace':[60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assingment_Completed':[1,2,1,3,2]
}
df = pd.DataFrame(data)
print(df)
```

```python
#1 line chart (trend)
plt.plot(df['Days'], df['Study_Hours'],marker='*')
plt.xlabel("Divij")
plt.ylabel("Days")
plt.title("Study Hours Over Days")
plt.show()
```

```python
#Advanced line chart
plt.figure(figsize=(7,4))
plt.plot(df['Days'], df['Study_Hours'],
         marker='o', linestyle='-', color='blue', label='Study Hours')

plt.plot(df['Days'], df['Marks'],
         marker='s',  color='red', label='Marks')

plt.xlabel('Days')
plt.ylabel('Value')
plt.title('Divij')
plt.legend()
```

```python
# Bar chart (comparison)
plt.bar(df['Days'], df['Marks'], color='green')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.title('DIVIJ')
plt.show()
```

```python
#Advanced bar graph
bars = plt.bar(df['Days'], df['Marks'], color='red')
for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, str(y),
             ha='center', va='bottom')

    plt.title('Divij')
    plt.xlabel('Days')
    plt.ylabel('Marks')
    plt.grid(axis='y')
```

```python
#3. histograms
data = {
    'Marks' : [50,60,55,70,75,80,85,90,45,68,72,77,82,88]
}
df = pd.DataFrame(data)
plt.hist(df['Marks'], bins=3, edgecolor='k',alpha=0.30)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Divij')
plt.show()
```

```python
import numpy as np
#Example data (more values for better histogram)
marks = [50,60,55,70,65,75,80,85,90,45,68,72,77,82,88]
#calcluate mean
mean_value = np.mean(marks)
plt.figure(figsize=(7,4))
plt.hist(marks, bins=6, color='b',alpha=0.30,edgecolor='k')
plt.axvline(mean_value, color='r', linestyle='dashed', linewidth=2,label='Mean')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Divij')
plt.legend()
plt.show()
```

```python
#scatter plot
plt.scatter(df['Study_Hours'],df['Marks'])
plt.title("Divij")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
df['Result'] = ['Pass' if marks >= 70 else 'Fail' for marks in df['Marks']]
colors = ['red' if result == 'Fail' else 'green' for result in df['Result']]
plt.scatter(df['Study_Hours'], df['Marks'], label='Marks',c=colors)
plt.title('Divij')
plt.xlabel('Study_Hours')
plt.ylabel('Marks')
plt.legend()
plt.show()
```

```python
#Dataset-2
data = {
    'Day': ['Mon','Tue','Wed','Thu','Fri','Sat','Sun','Mon','Tue','Wed'],
    'Region': ['North','South','East','West','North','South','East','West','North','South'],
    'Sales': [200,220,250,270,300,350,400,280,260,290],
    'Profit': [20,25,30,35,50,60,80,40,38,45],
    'Customers': [30,35,40,45,50,65,80,55,48,52],
    'Category': ['Food','Clothing','Electronics','Food','Clothing','Electronics','Food',
                 'Clothing','Electronics','Food']
}

df = pd.DataFrame(data)
print(df)
```

```python
#line plot
sns.lineplot(x='Day',y='Sales',data=df)
plt.title('Divij')
plt.show()
```

```python
#Bar plot
sns.barplot(x='Day',y='Profit',data=df)
plt.title("Divij")
plt.show()
```

```python
#histogram
sns.histplot(df['Sales'],bins=5)
plt.title("Divij")
plt.show()
```

```python
sns.scatterplot(x='Sales',y='Profit',data=df)
plt.title("Divij")
plt.show()
```
